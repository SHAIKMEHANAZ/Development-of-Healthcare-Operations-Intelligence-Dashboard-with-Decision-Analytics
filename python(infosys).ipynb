```python
def is_palindrome(text):
    text = text.lower().replace(" ", "")  # Convert to lowercase and remove spaces
    return text == text[::-1]

# Test cases
print(is_palindrome("madam"))
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("hello"))
```

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    else:
        list_fib = [0, 1]
        while len(list_fib) < n:
            next_fib = list_fib[-1] + list_fib[-2]
            list_fib.append(next_fib)
        return list_fib

# Test cases
print(fibonacci(10)) # First 10 Fibonacci numbers
print(fibonacci(1))  # First 1 Fibonacci number
print(fibonacci(0))  # 0 Fibonacci numbers
```

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# Test cases
print(f"Is 7 prime? {is_prime(7)}")
print(f"Is 1 prime? {is_prime(1)}")
print(f"Is 4 prime? {is_prime(4)}")
print(f"Is 23 prime? {is_prime(23)}")
print(f"Is 0 prime? {is_prime(0)}")
```

```python
def is_armstrong(number):
    # Convert number to string to easily get its digits and length
    num_str = str(number)
    n = len(num_str)

    sum_of_powers = 0
    for digit in num_str:
        sum_of_powers += int(digit) ** n

    return sum_of_powers == number

# Test cases
print(f"Is 153 an Armstrong number? {is_armstrong(153)}")
print(f"Is 9 an Armstrong number? {is_armstrong(9)}")
print(f"Is 370 an Armstrong number? {is_armstrong(370)}")
print(f"Is 371 an Armstrong number? {is_armstrong(371)}")
print(f"Is 407 an Armstrong number? {is_armstrong(407)}")
print(f"Is 1634 an Armstrong number? {is_armstrong(1634)}")
print(f"Is 10 an Armstrong number? {is_armstrong(10)}")
```

```python
def find_min_max(numbers):
    if not numbers:
        return None, None  # Return None for both if the list is empty

    min_val = numbers[0]
    max_val = numbers[0]

    for num in numbers:
        if num < min_val:
            min_val = num
        if num > max_val:
            max_val = num

    return min_val, max_val

# Test cases
list1 = [3, 1, 4, 1, 5, 9, 2, 6]
min1, max1 = find_min_max(list1)
print(f"List: {list1}, Min: {min1}, Max: {max1}")

list2 = [10]
min2, max2 = find_min_max(list2)
print(f"List: {list2}, Min: {min2}, Max: {max2}")

list3 = []
min3, max3 = find_min_max(list3)
print(f"List: {list3}, Min: {min3}, Max: {max3}")

list4 = [-5, -1, -10]
min4, max4 = find_min_max(list4)
print(f"List: {list4}, Min: {min4}, Max: {max4}")
```

### Program to Calculate Square Root

```python
import math

def calculate_square_root(number):
    if number < 0:
        return 'Cannot calculate square root of a negative number'
    return math.sqrt(number)

# Test cases
print(f"Square root of 9: {calculate_square_root(9)}")
print(f"Square root of 2: {calculate_square_root(2)}")
print(f"Square root of 0: {calculate_square_root(0)}")
print(f"Square root of -4: {calculate_square_root(-4)}")
```

## Question 2: Calculate Factorial (Numbers)
Problem Statement: Write a Python function to calculate the factorial of a non-negative integer.

Example Input:

n = 5
Expected Output:

120

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        res = 1
        for i in range(1, n + 1):
            res *= i
        return res

# Example usage:
print(factorial(5))
print(factorial(0))
```

## Question 5: Count Word Frequency (Dictionaries and Sets)
Problem Statement: Write a Python function to count the frequency of each word in a given sentence. The words should be case-insensitive.

Example Input:

sentence = "Hello world hello Python"
Expected Output:

{'hello': 2, 'world': 1, 'python': 1}

```python
def count_word_frequency(sentence):
    words = sentence.lower().split()
    word_counts = {}
    for word in words:
        word_counts[word] = word_counts.get(word, 0) + 1
    return word_counts

# Example usage:
print(count_word_frequency("Hello world hello Python"))
print(count_word_frequency("Apple banana Apple"))
```

## Question 8: Find Common Elements Between Two Lists (Lists and Arrays)
Problem Statement: Given two lists, list1 and list2, return a new list containing only the elements that are present in both lists. The order of elements in the output list does not matter.

Example Input:

list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
Expected Output:

[3, 4] # Order might vary

```python
def find_common_elements(list1, list2):
    return list(set(list1) & set(list2))

# Example usage:
print(find_common_elements([1, 2, 3, 4], [3, 4, 5, 6]))
print(find_common_elements(['apple', 'banana', 'orange'], ['orange', 'grape', 'banana']))
```

## Question 9: Check if Key Exists in Dictionary (Dictionaries and Sets)

Example Input:

my_dict = {'a': 1, 'b': 2, 'c': 3}
key_to_check = 'b'

True

## Question 10: Remove Duplicates from a List (Lists and Arrays / Sets)
Problem Statement: Given a list input_list, return a new list containing only the unique elements from the original list. The order of elements in the output list does not need to be preserved.

Example Input:

input_list = [1, 2, 2, 3, 4, 4, 5]
Expected Output:

[1, 2, 3, 4, 5] # Order might vary

```python
def remove_duplicates(input_list):
    return list(set(input_list))

# Example usage:
print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
print(remove_duplicates(['apple', 'banana', 'apple', 'orange']))
```
