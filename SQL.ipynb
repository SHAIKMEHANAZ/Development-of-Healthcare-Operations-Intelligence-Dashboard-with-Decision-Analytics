```python
import sqlite3

conn = sqlite3.connect("employee.db")
cursor = conn.cursor()

# Add DROP TABLE IF EXISTS to ensure a clean slate if the table already exists
cursor.execute("DROP TABLE IF EXISTS Employee;")

cursor.execute("""
CREATE TABLE Employee (
    Emp_ID INTEGER PRIMARY KEY,
    Emp_Name TEXT,
    Department TEXT,
    Job_Role TEXT,
    Salary REAL,
    Age INTEGER,
    City TEXT,
    Experience INTEGER
)
""")

conn.commit()
print("Table 'Employee' created successfully.")
```

```python
#1. SELECT — Display all employees
import sqlite3
import pandas as pd

conn = sqlite3.connect("employee.db")
cursor = conn.cursor()

df = pd.read_sql_query("SELECT * FROM Employee", conn)

display(df)
```

```python
#2. WHERE — Employees from IT
df = pd.read_sql_query("""
SELECT *
FROM Employee
WHERE Department = 'IT';
""", conn)

display(df)
```

```python
#3. Conditions — Salary greater than 50,000
df = pd.read_sql_query("""
SELECT Emp_Name, Salary
FROM Employee
WHERE Salary > 50000;
""", conn)

display(df)
```

```python
#4. AND — IT employees with salary above 50,000
df = pd.read_sql_query("""
SELECT Emp_Name, Department, Salary
FROM Employee
WHERE Department = 'IT'
AND Salary > 50000;
""", conn)

display(df)
```

```python
#5. LIKE — Names starting with S
df = pd.read_sql_query("""
SELECT *
FROM Employee
WHERE Emp_Name LIKE 'S%';
""", conn)

display(df)
```

```python
#6. Aggregate Functions — Salary calculation
df = pd.read_sql_query("""
SELECT
    COUNT(*) AS Total_Employees,
    SUM(Salary) AS Total_Salary,
    AVG(Salary) AS Average_Salary,
    MAX(Salary) AS Maximum_Salary,
    MIN(Salary) AS Minimum_Salary
FROM Employee;
""", conn)

display(df)
```

```python
#7. GROUP BY — Employees in each department
df = pd.read_sql_query("""
SELECT
    Department,
    COUNT(*) AS Employee_Count
FROM Employee
GROUP BY Department;
""", conn)

display(df)
```

```python
#8. HAVING — Departments with average salary above 50,000
df = pd.read_sql_query("""
SELECT
    Department,
    AVG(Salary) AS Average_Salary
FROM Employee
GROUP BY Department
HAVING AVG(Salary) > 50000;
""", conn)

display(df)
```

```python
#9. IN — Employees from Hyderabad or Chennai
df = pd.read_sql_query("""
SELECT Emp_Name, City
FROM Employee
WHERE City IN ('Hyderabad', 'Chennai');
""", conn)

display(df)
```

```python
#10. BETWEEN — Salary between 45,000 and 65,000
df = pd.read_sql_query("""
SELECT Emp_Name, Salary
FROM Employee
WHERE Salary BETWEEN 45000 AND 65000;
""", conn)

display(df)
```
